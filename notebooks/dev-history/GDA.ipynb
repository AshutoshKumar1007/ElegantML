Normalizing each feature (zero mean and 1 std)

```python
x = df.drop('target',axis= 1)
x = (x - x.mean())/x.std()
y = df.target
```

```python
x_train, x_test = x[0:850], x[851:]
y_train, y_test = y[0:850], y[851:]
x_train.shape
```

```python
indicator = lambda y,cla_ss: 1 if y == cla_ss else 0
count_pos = 0
count_neg = 0
for i in range(x_train.shape[0]):
  if(indicator(y_train[i],1)):
    count_pos += 1
count_neg = x_train.shape[0] - count_pos
count_neg,count_pos

```

```python
m,n = x_train.shape
mu1 = np.zeros(n)
mu2 = np.zeros(n)
for i in range(m):
  mu1 += indicator(y_train[i],1)*x_train.iloc[i]
  mu2 += indicator(y_train[i],0)*x_train.iloc[i]
mean1 = np.array(mu1/count_pos)
mean0 = np.array(mu2/count_neg)
mean1,mean0
phi = count_pos/m
phi_not = count_neg/m
mean1,mean0,phi
mu = [mean0,mean1]
mu
mu = np.array(mu)
mu


```

```python
sigma = np.zeros((n,n))
for i in range(x_train.shape[0]):
  sigma += np.outer((x_train.iloc[i] - mu[y_train[i]]),(x_train.iloc[i] - mu[y_train[i]]))
sigma = sigma/m
sigma
```

```python
np.allclose(sigma,sigma.T)
```

```python
def gausian(Class,i):
  return ((((2*np.pi)**(n/2))*np.linalg.det(sigma)**0.5)**(-1))*np.exp((-1/2)*(x.iloc[i] - mu[Class])@(np.linalg.inv(sigma)@(x.iloc[i] - mu[Class])))
```

```python
gausian(0,500)
```

```python
def hypothesis(i):
  p_of_y_x = []
  p_of_y_x.append(gausian(0,i)*(1 - phi))
  p_of_y_x.append(gausian(1,i)*phi)
  return p_of_y_x

```

```python
hypothesis(1)
```

```python
def accuracy():
  count = 0
  for i in range(1000):
    if np.argmax(hypothesis(i)) == y[i]:
      count += 1
  return count


```
