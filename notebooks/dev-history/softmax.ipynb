````python
w = np.random.normal(loc = 0.0, scale = 1.0, size = (6,34))

w.shape,x_train.shape
````

````python
def hypothesis(x_train,w,i):
  h_theta_x  = np.zeros(6)
  for k in range(6):
    h_theta_x[k] += np.exp(np.dot(w[k], x_train.loc[i]))
  sum = np.sum(h_theta_x)
  return h_theta_x/sum
````

````python
def log_likelihood(x_train,y_train,w):
  l_theta = 0
  indicator = lambda y,l: 1 if y == l else 0
  epsilon = 1e-15
  m,n = x_train.shape
  k = w.shape[0]
  for i in range(m):
    h_theta = hypothesis(x_train,w,i)
    for j in range(k):
      l_theta = l_theta + (np.log(h_theta[j] + epsilon )*indicator(y_train[i],j))
  return l_theta


````

````python
l = log_likelihood(x_train,y_train,w)
l
````

````python
def derivative1(x_train,y_train,w):
  m,n = x_train.shape
  K = w.shape[0]
  indicator = lambda y,l: 1 if y == l else 0
  dl_dw = np.zeros((w.shape))
  for i in range(m):
    h_theta = hypothesis(x_train,w,i)
    for j in range(K):
      h_theta_j = h_theta[j]
      indic_y_J = indicator(y_train[i],j)
      for k in range(n):
        dl_dw[j,k] += (indic_y_J - h_theta_j)*x_train.iloc[i,k]
  return dl_dw
````

````python
der1= derivative1(x_train,y_train,w)
der1
````

````python
def derivative(x_train,y_train,w):
  m,n = x_train.shape
  K = w.shape[0]
  indicator = lambda y,l: 1 if y == l else 0
  dl_dw = np.zeros((w.shape))
  for j in range(K):
    for k in range(n):
      der = 0
      for i in range(m):
        h_theta = hypothesis(x_train,w,i)
        der += (indicator(y_train[i],j) - h_theta[j])*x_train.iloc[i,k]
      dl_dw[j,k] = der
  return dl_dw
````

````python
def soft_max(x_train,y_train,w,alpha,no_iters):
  hist_log_likelihood =[]
  for i in range(no_iters):
    dL_dw = derivative1(x_train,y_train,w)
    w += alpha*dL_dw
    hist_log_likelihood.append(log_likelihood(x_train,y_train,w))
    if i % math.ceil(no_iters/10)== 0:
      print(f"{i}th iteration ::  log_likelihood : {hist_log_likelihood[-1]}")
  return w,hist_log_likelihood[-1]
````
