```python
import numpy as np
import pandas as pd

import math

A = pd.read_csv('/content/house_data.csv')
A
```

```python
x= A.drop(["Price$"],axis = 1)
x['xo'] = 1
x
```

```python
y = A['Price$']
y = y.to_numpy()
type(y)
```

```python
w = np.random.rand(5)
type(w)
```

```python
def hypothesis(X,W):
  return np.dot(X,W)
```

```python
def cost_function(x,y,w):
  temp = 0
  for i in range(len(y)):
    temp += (y[i] - hypothesis(x.loc[i].values, w))**2
  return temp/2

```

```python
z = cost_function(x,y,w)
x.iloc[1,1]
```

```python
def derivative(x,y,w):
  m,n = x.shape
  dJ_dw = np.zeros(n)
  for i in range(m):
    error = hypothesis(x.loc[i].values,w) - y[i]
    for j in range(n):
      dJ_dw[j] += error*(x.iloc[i,j])
  return dJ_dw/m
```

```python
derivative(x,y,w)
```

```python
def batch_gradient(x,y,w,alpha,no_iters):
  cost_hist =[]
  for i in range(no_iters):
    dj_dw = derivative(x,y,w)
    w = w - alpha*dj_dw
    if( i < 10000):
      cost_hist.append(cost_function(x,y,w))
    if i%math.ceil(no_iters/10)==0:
      print(f"iteration {i:4d} : Cost {cost_hist[-1]:8.2f}")
  return w,cost_hist



```

```python
def shocastic_gradient(x,y,w,alpha,no_iters):
  m,n = x.shape
  cost_hist = []
  for i in range(no_iters):
    for j in range(n):
      for k in range(m):
        error = (hypothesis(x.loc[k].values,w) - y[k])*x.iloc[k,j]
        w[j] = w[j] - (alpha*error)
    if i < 10000:
      cost_hist.append(cost_function(x,y,w))
    if i%math.ceil(no_iters/10)==0:
      print(f"iteration: {i:4d} : Cost {cost_hist[-1]:8.2f}")
  return w,cost_hist
```

```python
no_iters = 10000
alpha = 0.0000005
w = np.zeros(5)
w_final, cost_final = shocastic_gradient(x,y,w,alpha, no_iters)
print(f"optimised value of feauters: {w_final}")
```
