```python
import numpy as np
import pandas as pd
import matplotlib as mpl
from matplotlib import pyplot as plt
```

```python

df = pd.read_csv('dataset\heart.csv')
shuffled = df.sample(frac=1.0, replace=False)
df = shuffled.reset_index(drop = True)
df['x0'] = 1
df.head(20)
```

```python
x = df.drop(['target'], axis= 1)
y = df.target
y
```

```python
x_train, x_test = x[0:240], x[241:]
y_train, y_test = y[0:240], y[241:]
x_test.head(30)
```

```python
w = np.random.randn(x_train.shape[1])
w
```

```python
def hypothesis(X,w):
  z = np.dot(X.T,w)
  return 1/(1 + np.exp(-z))
```

```python
hypothesis(x_train.iloc[1],w)
```

```python
m,n = x_train.shape
h_test = []
for i in range(m):
  h_test.append(hypothesis(x_train.iloc[i],np.ones(n)))

```

```python
def log_likelihood(x,y,w):
  m,n = x.shape
  l_0 = 0
  epsilon = np.exp(-15)
  for i in range(m):
    l_0 += (y[i]*np.log(hypothesis(x_train.iloc[i],w) + epsilon) + (1 - y[i])*np.log(1 - hypothesis(x_train.iloc[i],w) + epsilon) )
  return l_0

```

```python
def derivative(x,y,w):
  m,n = x.shape
  dL_dw = np.zeros(n)
  for i in range(m):
    error = y[i] - hypothesis(x.loc[i].values,w)
    for j in range(n):
      dL_dw[j] += error* (x.iloc[i,j])
  return dL_dw
```

```python
der = derivative(x_train,y_train,w)
```

```python
def derivative1(x, y, w):
    m, n = x.shape
    dL_dw = np.zeros(n)

    for j in range(n):
        error = 0.0
        for i in range(m):
            #print(y[i])
            #print(hypothesis(x.iloc[i].values, w))
            error += (y[i] - hypothesis(x.iloc[i].values, w)) * x.iloc[i, j]
        dL_dw[j] = error
    return dL_dw

```

```python
import math
```

```python
def logistic(x,y,w,alpha,no_iters):
  log_like_hist = [0]
  m,n = x.shape
  for i in range(no_iters):
    dl_dw = derivative1(x,y,w)
    w += dl_dw*alpha
    
    if( i < 10000):
      log_like_hist.append(log_likelihood(x,y,w))
    if i%math.ceil(no_iters/10)==0:
      print(f"iteration {i:4d} : loglikelihood {log_like_hist[-1]:8.2f}")
  return w,log_like_hist
```

```python
alpha = 0.0001
no_iters = 100
w_final, hist_final = logistic(x_train,y_train,w,alpha,no_iters)
w_final
```

```python
hypothesis(x_train.iloc[123],w_final)
```

```python
def accuracy(x_test,w_final):
  count = 0
  for i in range(241,270):
    if round(hypothesis(x_test.iloc[i],w_final)) == y_test[i]:
      count += 1
  return count


```
